```python
import pandas as pd
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

# --------------------------------------------------
# 1. LOAD DATASET
# --------------------------------------------------

df = pd.read_csv("IMDB_Dataset_CLEANED.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

print("\nFirst 5 rows:")
print(df.head())


# --------------------------------------------------
# 2. CHECK DATASET
# --------------------------------------------------

print("\nDataset Information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# --------------------------------------------------
# 3. REMOVE MISSING VALUES
# --------------------------------------------------

df = df.dropna()

print("\nAfter removing missing values:")
print(df.shape)


# --------------------------------------------------
# 4. REMOVE DUPLICATES
# --------------------------------------------------

df = df.drop_duplicates()

print("\nAfter removing duplicates:")
print(df.shape)


# --------------------------------------------------
# 5. STOPWORDS
# --------------------------------------------------

stop_words = set(stopwords.words("english"))


# --------------------------------------------------
# 6. TEXT CLEANING FUNCTION
# --------------------------------------------------

def clean_review(text):

    # Convert to string
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Join words
    return " ".join(tokens)


# --------------------------------------------------
# 7. CLEAN REVIEWS
# --------------------------------------------------

df["cleaned_review"] = df["review"].apply(clean_review)


# --------------------------------------------------
# 8. DISPLAY RESULTS
# --------------------------------------------------

print("\nOriginal vs Cleaned Reviews\n")

for i in range(5):

    print("ORIGINAL:")
    print(df.iloc[i]["review"])

    print("\nCLEANED:")
    print(df.iloc[i]["cleaned_review"])

    print("\n" + "=" * 80)


# --------------------------------------------------
# 9. SENTIMENT DISTRIBUTION
# --------------------------------------------------

print("\nSentiment Distribution:")
print(df["sentiment"].value_counts())


# --------------------------------------------------
# 10. SAVE CLEANED DATASET
# --------------------------------------------------

df.to_csv(
    "final_cleaned_imdb_reviews.csv",
    index=False
)

print("\nFinal cleaned dataset saved successfully!")
```
